```python
from pathlib import Path
import json
```

```python
file = Path("/home/studyztp/test_ground/experiments/nugget-eval/gem5-experiments/nugget-protocol-NPB/experiments/1-thread-dynamorio-count-nugget/bt-1853/marker_log.txt")
```

```python
with open(file, "r") as f:
    lines = f.readlines()
all_lines = "".join(lines)
orders = all_lines.split("->")
```

```python
cut_down_index = 0
counter = 0
for index, i, in enumerate(orders):
    if i == "0":
        counter += 1
    if counter == 2861:
        cut_down_index = index

orders = orders[:cut_down_index + 1]

counter_1 = 0
counter_2 = 0
counter_3 = 0

printed = False

for i in orders:
    if i == "1":
        counter_1 += 1
    elif i == "2":
        counter_2 += 1
    elif i == "0":
        counter_3 += 1
    if counter_1 == 26 and not printed:
        print(counter_1)
        print(counter_2)
        print(counter_3)
        print("---")
        counter_2 = 0
        counter_3 = 0
        counter_1 = 0
        printed = True
    if counter_2 == 1503 and printed:
        print(counter_1)
        print(counter_2)
        print(counter_3)
        print("---")
        break
    
```

```python
json_file = Path("/home/studyztp/test_ground/experiments/nugget-eval/gem5-experiments/looppoint_analysis_of_x86_bt_A_1853.json")

with open(json_file, "r") as f:
    json_data = json.load(f)

sim_region = json_data["1"]

bb_inst_map = sim_region["bb_inst_map"]
bbv = sim_region["global_bbv"]

user_inst = 0
lib_inst = 0

for pc, num_inst in bb_inst_map.items():
    if "0x7ffff7" in pc:
        lib_inst += num_inst * bbv[pc]
    else:
        user_inst += num_inst * bbv[pc]

print("User Instructions:", user_inst)
print("Library Instructions:", lib_inst)
```
